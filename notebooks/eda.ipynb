```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data_path = "../data/processed/train"
classes = sorted(os.listdir(data_path))

class_counts = {}
for cls in classes:
    count = len(os.listdir(os.path.join(data_path, cls)))
    class_counts[cls] = count

df = pd.DataFrame(list(class_counts.items()), columns=['Class', 'Count'])

plt.figure(figsize=(12, 10))
sns.barplot(data=df, y='Class', x='Count', palette='viridis')
plt.title('PlantVillage Training Class Distribution (38 Classes)')
plt.xlabel('Number of Images')
plt.ylabel('Crop & Disease Class')
plt.tight_layout()
plt.savefig('eda_class_distribution.png')
plt.show()

print(f"Total Training Images: {df['Count'].sum()}")
print(f"Min Class Count: {df['Count'].min()} | Max Class Count: {df['Count'].max()}")
```
